```python
class ActivityMLP(nn.Module):

    def __init__(self,
                 input_dim=40,
                 hidden_dims=(256, 128, 64),
                 num_classes=12,
                 activation="relu",
                 dropout=0.2,
                 batch_norm=True):
        super().__init__()

        activations = {"relu": nn.ReLU, "tanh": nn.Tanh, "gelu": nn.GELU}

        layers = []
        prev = input_dim
        for h in hidden_dims:
            layers.append(nn.Linear(prev, h))
            if batch_norm:
                layers.append(nn.BatchNorm1d(h))
            layers.append(activations[activation]())
            if dropout > 0:
                layers.append(nn.Dropout(dropout))
            prev = h
        layers.append(nn.Linear(prev, num_classes))

        self.net = nn.Sequential(*layers)

    def forward(self, X):
        return self.net(X)

    def evaluate(self, dl, loss_fn, device="cpu"):
        self.eval()
        total_loss, correct, n = 0.0, 0, 0
        with torch.no_grad():
            for xb, yb in dl:
                xb, yb = xb.to(device), yb.to(device)
                logits = self(xb)
                total_loss += loss_fn(logits, yb).item() * len(yb)
                correct += (logits.argmax(1) == yb).sum().item()
                n += len(yb)
        return total_loss / n, correct / n

    def fit(self, train_dl, epochs=30, lr=1e-3, weight_decay=1e-4,
            class_weights=None, val_dl=None, device="cpu"):
        self.to(device)
        weight = None if class_weights is None else class_weights.to(device)
        loss_fn = nn.CrossEntropyLoss(weight=weight)
        optimizer = torch.optim.Adam(self.parameters(), lr=lr, weight_decay=weight_decay)

        self.loss_history = []
        self.val_loss_history = []
        self.val_acc_history = []

        for epoch in range(epochs):
            self.train()
            total_loss, n = 0.0, 0
            for xb, yb in train_dl:
                xb, yb = xb.to(device), yb.to(device)
                optimizer.zero_grad()
                loss = loss_fn(self(xb), yb)
                loss.backward()
                optimizer.step()
                total_loss += loss.item() * len(yb)
                n += len(yb)
            self.loss_history.append(total_loss / n)

            if val_dl is not None:
                val_loss, val_acc = self.evaluate(val_dl, loss_fn, device)
                self.val_loss_history.append(val_loss)
                self.val_acc_history.append(val_acc)

        return self

    def predict(self, X, device="cpu"):
        self.eval()
        with torch.no_grad():
            return self(X.to(device)).argmax(1).cpu()
```
